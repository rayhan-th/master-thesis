# Low-Flow (Fishing / Ecological) Release Curve — Neuer Teich

**Purpose:** Plots the storage–discharge relationship for the ecological
(low-flow / fishing) release rule at Neuer Teich. This operating rule
is used alongside the spillway curve in TALSIM-NG.

**What it does:**
- Defines paired storage (Tsd m³) and release (m³/s) values for the
  low-flow outlet
- Produces a publication-ready plot

**Input:** Hard-coded storage–release table  
**Output:** Release curve plot

---

```python
import numpy as np
import matplotlib.pyplot as plt

# =============================================================================
# DATA  (Storage in Tsd m³, Release in m³/s)
# =============================================================================
storage = [
    0.000, 1.457, 9.880, 20.689, 32.638,
    49.235, 70.526, 98.365, 132.735, 175.071,
    227.465, 276.831, 334.823, 342.833, 380.000,
    412.648, 430.000
]

release = [
    0.00000, 0.00000, 0.00000, 0.00000, 0.31000,
    0.32000, 0.33000, 0.34000, 0.35000, 0.36000,
    0.37000, 0.39000, 0.40000, 0.00000, 0.00000,
    0.00000, 0.00000
]

storage = np.array(storage)
release = np.array(release)

# =============================================================================
# DATA 2 - Second fishing curve
# =============================================================================
storage2 = [
    0.000, 0.017, 0.258, 0.441, 1.455,
    4.297, 11.198, 18.000, 23.043, 33.057,
    41.116, 55.480, 79.302, 93.396, 126.347,
    155.541, 200.949, 240.209, 284.225, 300.000
]

release2 = [
    0.00000, 0.00000, 0.00000, 0.00000, 0.00000,
    0.00000, 0.00000, 0.00000, 0.34700, 0.35200,
    0.36000, 0.37000, 0.37500, 0.38500, 0.39200,
    0.40200, 0.41000, 0.41700, 0.00000, 0.00000
]

storage2 = np.array(storage2)
release2 = np.array(release2)

# =============================================================================
# PLOT
# =============================================================================
fig, ax = plt.subplots(figsize=(6.30, 3.20))

# Line
ax.plot(storage, release, color="#1f77b4", linewidth=1.0,
        zorder=2, label="Neuer Teich")

# Fill below the curve — reservoir style
ax.fill_between(storage, release, 0,
                color="cyan", alpha=0.25, zorder=1)

# Second curve
ax.plot(storage2, release2, color="#e63946", linewidth=1.0,
        zorder=2, label="Mittelteich")
ax.fill_between(storage2, release2, 0,
                color="#e63946", alpha=0.15, zorder=1)

# Axes labels and formatting
ax.set_xlabel("Storage [Tsd m³]", fontsize=6)
ax.set_ylabel("Release [m³/s]", fontsize=6)


ax.tick_params(axis="both", which="major", labelsize=5.5, direction="out")
ax.tick_params(axis="both", which="minor", length=2, width=0.5,
               color="0.5", direction="out")

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_linewidth(0.6)
ax.spines["bottom"].set_linewidth(0.6)

ax.grid(False)
# Minimum storage markers
min_storage1 = 20.689
ax.axvline(x=min_storage1, color="#1f77b4", linewidth=0.8, linestyle="--",
           zorder=4, label="_nolegend_")
ax.text(min_storage1 + 1, ax.get_ylim()[1] * 0.95,
        "Min. Storage Neuer_Teich\n " + str(min_storage1) + " Tsd m³",
        fontsize=5, color="#1f77b4", va="top", ha="left")

min_storage2 = 18.000
ax.axvline(x=min_storage2, color="#e63946", linewidth=0.8, linestyle="--",
           zorder=4, label="_nolegend_")
ax.text(min_storage2 + 1, ax.get_ylim()[1] * 0.80,
        "Min. Storage Mittelteich\n   " + str(min_storage2) + " Tsd m³",
        fontsize=5, color="#e63946", va="top", ha="left")

ax.legend(fontsize=5.5, loc="lower right", framealpha=0.8,
          borderpad=0.4, labelspacing=0.3,
          handlelength=1.5, handletextpad=0.4,
          bbox_to_anchor=(1.0, 0.1))  # ← increase 0.15 to move higher
# X-axis ticks every 25 from 0 to 430
ax.set_xticks(np.arange(0, 431, 25))
ax.tick_params(axis="x", which="major", labelsize=5.0, rotation=45)

plt.tight_layout()
plt.savefig("storage_release_curve.png", dpi=300, bbox_inches="tight")
print("Saved → storage_release_curve.png")
plt.show()
```
