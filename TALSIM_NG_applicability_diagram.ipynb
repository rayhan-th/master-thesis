# TALSIM-NG Applicability — Sankey Diagram

**Purpose:** Creates a Sankey (alluvial) diagram that visualises how the five
study objectives connect to the three modelling frameworks
(TALSIM-NG, J2000, LUTZ), summarising the methodological scope of the thesis.

**What it does:**
- Defines nodes for each study objective and each model
- Assigns flow values and colours representing the strength of the link
- Renders an interactive Plotly Sankey figure and exports it to PNG via Kaleido

**Input:** Hard-coded node/link data  
**Output:** Interactive Sankey HTML + static PNG

```python
import plotly.graph_objects as go

nodes = [
    "Setup & Data",
    "Continuous Simulation & Validation",
    "Reservoir Operation",
    "Event-Based Simulation",
    "Operational Feasibility",
    "TALSIM-NG",
    "J2000",
    "LUTZ"
]

source = [0,0,0, 1,1,1, 2,2,2, 3,3,3, 4,4,4]
target = [5,6,7, 5,6,7, 5,6,7, 5,6,7, 5,6,7]
value  = [2,3,3, 3,2,1, 3,1,1, 2,3,3, 2,3,3]

link_colors = []
for v in value:
    if v == 3:
        link_colors.append("rgba(59,109,17,0.40)")    # deep forest green
    elif v == 2:
        link_colors.append("rgba(139,90,43,0.45)")    # earthy warm brown
    else:
        link_colors.append("rgba(180,30,30,0.35)")    # clear red

fig = go.Figure(go.Sankey(
    arrangement="snap",
    node=dict(
        pad=25,
        thickness=35,
        line=dict(color="white", width=0.5),
        label=nodes,
        color=[
            "#5F5E5A",
            "#5F5E5A",
            "#5F5E5A",
            "#5F5E5A",
            "#5F5E5A",
            "#3C3489",
            "#0C447C",
            "#085041"

        ],
        x=[0.01, 0.01, 0.01, 0.01, 0.01,   # clusters — left column
           0.99, 0.99, 0.99],               # models — right column
        y=[0.10, 0.30, 0.50, 0.70, 0.90,   # clusters — evenly spaced
           0.20, 0.55, 0.88]               # models — evenly spaced
        
    ),
    link=dict(
        source=source,
        target=target,
        value=value,
        color=link_colors
    )
))

fig.update_layout(

    font=dict(size=14, family="monospace", color="#2C2C2A"),
    width=750,
    height=480,
    paper_bgcolor="white",
    margin=dict(l=30, r=30, t=60, b=60),
    annotations=[
    # --- axis labels ---
        dict(
            x=-0.045, y=0.5,
            xref="paper", yref="paper",
            showarrow=False,
            text="Evaluation Criteria",
            font=dict(size=12, family="monospace", color="#5F5E5A"),
            textangle=-90,
            align="center"
        ),
        dict(
            x=1.045, y=0.5,
            xref="paper", yref="paper",
            showarrow=False,
            text="Model (s) Used",
            font=dict(size=12, family="monospace", color="#5F5E5A"),
            textangle=-90,
            align="center"
        ),
    # legend title
        dict(
            x=0, y=-0.08,
            xref="paper", yref="paper",
            showarrow=False,
            text="<b> Applicability:</b>",
            font=dict(size=11, family="monospace", color="#2C2C2A"),
            align="left"
             
        ),
        dict(
            x=0.01, y=-0.13,
            xref="paper", yref="paper",
            showarrow=False,
            text='<span style="color:#3B6D11; font-size:16px;">&#9632;</span>  Strong',
            font=dict(size=11, family="monospace", color="#2C2C2A"),
            align="left"
        ),
        dict(
            x=0.13, y=-0.13,
            xref="paper", yref="paper",
            showarrow=False,
            text='<span style="color:#8B5A2B; font-size:16px;">&#9632;</span>  Moderate',
            font=dict(size=11, family="monospace", color="#2C2C2A"),
            align="left"
        ),
        dict(
            x=0.38, y=-0.13,
            xref="paper", yref="paper",
            showarrow=False,
            text='<span style="color:#B41E1E; font-size:16px;">&#9632;</span>   Limited or not applicable',
            font=dict(size=11, family="monospace", color="#2C2C2A"),
            align="left"
        ),
    ]
)
fig.write_image("sankey_evaluation.png", scale=4, width=750, height=480)
fig.show(renderer="notebook")
```
