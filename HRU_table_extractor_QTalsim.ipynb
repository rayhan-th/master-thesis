# HRU Table Extractor — QTalsim → Excel

**Purpose:** Converts the pipe-delimited `.EFL` file produced by QTalsim into a
clean, human-readable Excel table for inspection and documentation.

**What it does:**
- Reads the `.EFL` file and filters rows that start with `| A` (HRU data lines)
- Splits each row into 15 named columns:
  `EZG`, `Slope_%`, `Area_%`, `SoilID`, `LandUseID`, `IrrID`,
  `CN_ID`, `Type`, `Stream`, `Elevation_m`, `HRU_ID`,
  `OutID`, `InletID`, `OutletID`, `Branch`
- Writes the result to an `.xlsx` file next to the source `.EFL`

**Input:** `ZR_test1.EFL`  
**Output:** `ZR_test1.xlsx`

---

```python
import pandas as pd
```

```python
# === Step 1: Read the .EFL file ===
file_path = r"C:\Users\raah\Desktop\Aufgabe\TALSIM_NG\EZG\Ziegenrueck\QTalsim_ZR\HRU generation\ZR_test1.EFL"

with open(file_path, 'r', encoding='utf-8') as f:
    lines = f.readlines()

# === Step 2: Extract data lines (those that start with "| A") ===
data_lines = [line.strip() for line in lines if line.strip().startswith('| A')]

# === Step 3: Parse each line into columns ===
parsed_rows = []
for line in data_lines:
    parts = line.strip('|').split('|')
    if len(parts) >= 3:
        left = parts[0].split()
        middle = parts[1].split()
        right = parts[2].split()
        row = left + middle + right

        # Pad row to make sure it has exactly 14 elements
        if len(row) < 15:
            row += [''] * (15 - len(row))
        elif len(row) > 15:
            row = row[:15]  # Trim if too long

        parsed_rows.append(row)

# === Step 4: Define column headers ===
columns = [
    'EZG', 'Gef_%', 'Flaeche_%', 'Bod_ID', 'Lnz_ID', 'Irr_ID',
    'CN_ID', 'Type', 'Stream', 'Elev_m', 'HRU_ID', 'Out_ID',
    'Inlet_ID', 'Outlet_ID', 'Branch'
]

# === Step 5: Create DataFrame and save to Excel ===
df = pd.DataFrame(parsed_rows, columns=columns)
output_file = file_path.replace('.EFL', '.xlsx')
df.to_excel(output_file, index=False)

print(f"✅ Data successfully written to: {output_file}")
```
