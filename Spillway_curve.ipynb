# Spillway Release Curve — Neuer Teich

**Purpose:** Plots the storage–discharge relationship for the Neuer Teich
spillway. This curve is used as an operating rule in the TALSIM-NG
reservoir element.

**What it does:**
- Defines paired storage (Tsd m³) and release (m³/s) values
- Marks the spillway activation threshold (334.824 Tsd m³)
- Produces a publication-ready plot saved to `spillway_curve.png`

**Input:** Hard-coded storage–release table  
**Output:** `spillway_curve.png`

---

```python
import numpy as np
import matplotlib.pyplot as plt

# =============================================================================
# DATA - Spillway curve (Neuer Teich)
# =============================================================================
storage_sp = [
    0.000, 334.824, 342.244, 349.780, 357.433,
    365.205, 373.095, 381.107, 389.239, 397.494,
    405.873, 414.376, 423.005, 430.000
]

release_sp = [
    0.00000, 0.00000, 0.02000, 0.03000, 0.08000,
    0.15000, 0.23000, 0.34000, 0.46000, 0.60000,
    0.75000, 0.93000, 1.12000, 1.29000
]

storage_sp = np.array(storage_sp)
release_sp = np.array(release_sp)

# =============================================================================
# PLOT
# =============================================================================
fig, ax = plt.subplots(figsize=(6.30, 3.20))

# Spillway curve line + fill
ax.plot(storage_sp, release_sp, color="#2ca02c", linewidth=1.0,
        zorder=2, label="Neuer Teich Spillway")
ax.fill_between(storage_sp, release_sp, 0,
                color="cyan", alpha=0.25, zorder=1)

# Spillway activation marker
spillway_start = 334.824
ax.axvline(x=spillway_start, color="red", linewidth=0.8,
           linestyle="--", zorder=4, label="_nolegend_")
ax.text(spillway_start + 2, ax.get_ylim()[1] * 0.95,
        "Spillway activation\n" + str(spillway_start) + " Tsd m³",
        fontsize=5, color="red", va="top", ha="left")

# Axes labels
ax.set_xlabel("Storage [Tsd m³]", fontsize=6)
ax.set_ylabel("Release [m³/s]", fontsize=6)


# X ticks every 25
ax.set_xticks(np.arange(0, 431, 25))
ax.tick_params(axis="x", which="major", labelsize=5.0, rotation=45)
ax.tick_params(axis="y", which="major", labelsize=5.5, direction="out")
ax.tick_params(axis="both", which="minor", length=2, width=0.5,
               color="0.5", direction="out")

# Spines
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_linewidth(0.6)
ax.spines["bottom"].set_linewidth(0.6)

ax.grid(False)

ax.legend(fontsize=5.5, loc="upper left", framealpha=0.8,
          borderpad=0.4, labelspacing=0.3,
          handlelength=1.5, handletextpad=0.4)

plt.tight_layout()
plt.savefig("spillway_curve.png", dpi=300, bbox_inches="tight")
print("Saved → spillway_curve.png")
plt.show()
```
