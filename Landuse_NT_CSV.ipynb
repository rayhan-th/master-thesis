# Land Use Parameter Export

**Purpose:** Encodes land-use attributes for the five land-cover classes found in the
Ziegenrück / Neuer Teich catchment and writes them to a CSV that can be loaded
directly into TALSIM-NG.

**What it does:**
- Defines root depth (m), plant coverage (%), and leaf area index for:
  `Ackerland`, `Grünland`, `Wald`, `Siedlung`, `Gewässer`
- Writes the table to `land_use_data_ZR.csv`

**Input:** Hard-coded values from GIS / task data  
**Output:** `land_use_data_ZR.csv`

---

```python
import csv
```

```python
#This data frame has been created by directly using the data from TASK2

data = [
    ["Name", "Root_depth_m", "Plant_coverage_%", "Leaf_Area_Index"],
    ["Ackerland", 1.2, 80, 1.3],
    ["Gruenland", 0.9, 100, 1], 
    ["Wald", 1.6, 100, 1],
    ["Siedlung", 0.7, 10, 1],
    ["Gewaesser", 10, 100, 1],
]

with open("land_use_data_ZR.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)
```
